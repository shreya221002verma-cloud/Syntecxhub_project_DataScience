```python
data = {
    'Date': pd.date_range(start='2023-01-01', periods=12, freq='ME'),
    'Category': ['Electronics', 'Clothing', 'Groceries'] * 4,
    'Sales': [12000, 8000, 6000, 15000, 9000, 7000,
              17000, 9500, 7500, 20000, 11000, 8500]
}
```

```python
df = pd.DataFrame(data)
df['Date'] = pd.to_datetime(df['Date'])
```

```python
monthly_sales = df.groupby('Date')['Sales'].sum()
```

```python
plt.figure()
monthly_sales.plot(marker='o')
plt.title('Sales Over Time')
plt.xlabel('Date')
plt.ylabel('Total Sales')
plt.grid(True)

```

```python
category_sales = df.groupby('Category')['Sales'].sum()

plt.figure()
category_sales.plot(kind='bar')
plt.title('Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')
```

```python
plt.figure()
category_sales.plot(kind='pie', autopct='%1.1f%%', startangle=90)
plt.title('Category Share in Sales')
plt.ylabel('')
```

```python
import numpy as np
```

```python

data = {
    'Sales': [200, 250, 300, 350, 400, 450, 500, 550, 600, 650],
    'Marketing_Spend': [50, 60, 70, 80, 90, 100, 110, 120, 130, 140],
    'Profit': [20, 30, 45, 60, 80, 100, 120, 140, 160, 180],
    'Discount': [5, 7, 6, 8, 10, 12, 11, 13, 15, 14]
}

df = pd.DataFrame(data)
```

```python

corr = df.corr(method='pearson')
```

```python
mask = np.triu(np.ones_like(corr, dtype=bool))
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(corr, mask=mask, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
```

```python
sns.pairplot(df)
```

```python
data = {
    'Region': ['A'] * 15 + ['B'] * 15,
    'Sales': [200, 220, 250, 270, 300, 320, 350, 370, 400, 420, 450, 480, 500, 520, 600,
              180, 190, 210, 230, 250, 260, 280, 300, 310, 330, 350, 360, 380, 400, 900]
}

df = pd.DataFrame(data)
```

```python
plt.figure()
sns.histplot(df['Sales'], kde=True)
plt.title('Sales Distribution (Histogram + KDE)')
plt.xlabel('Sales')
plt.ylabel('Frequency')
```

```python
plt.figure()
sns.boxplot(x='Region', y='Sales', data=df)
plt.title('Boxplot of Sales by Region')
plt.savefig('boxplot_region.png')
```

```python
plt.figure()
sns.kdeplot(df[df['Region'] == 'A']['Sales'], label='Region A')
sns.kdeplot(df[df['Region'] == 'B']['Sales'], label='Region B')
plt.title('KDE Comparison of Sales')
plt.xlabel('Sales')
plt.legend()
```
